```python
# AIMarket School — Pins on the Live Map
# Cell 1: GAIA health + attested weather pin (Open-Meteo relay ≠ in-situ station).
import urllib.request, json

GAIA = 'https://iot.modelmarket.dev'

def post(url, body):
  req = urllib.request.Request(url, data=json.dumps(body).encode(), headers={'content-type': 'application/json'})
  with urllib.request.urlopen(req, timeout=25) as r:
    return json.load(r)

health = json.load(urllib.request.urlopen(GAIA + '/health', timeout=20))
print('GAIA devices', health.get('devices'), 'sim_control', health.get('sim_control'))
wx = post(GAIA + '/ai-market/v2/invoke', {'capability_id': 'gaia.weather.read@v1', 'input': {'device_id': 'om-wx-01'}})
reading = (wx.get('output') or {}).get('reading') or {}
print('device', reading.get('device_id'), 'model', reading.get('model'))
print('values', reading.get('values'))
print('attestation', ((wx.get('output') or {}).get('attestation') or {}).get('algorithm'))
assert wx.get('ok'), 'expected LIVE weather ok'
```

```python
# Cell 2: ATLAS health — planetary map over GAIA.
atlas = json.load(urllib.request.urlopen('https://atlas.modelmarket.dev/health', timeout=20))
print('ATLAS stations', atlas.get('stations'), 'gaia', atlas.get('gaia'))
print('Academy: licence gates, honesty claims, add-sensor Recipe A')
```
