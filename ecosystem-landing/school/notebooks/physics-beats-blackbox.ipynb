```python
# AIMarket School — Physics Beats the Black Box
# Cell 1: discover physics-shaped oracles on the live hub.
import urllib.request, json

def search(intent, limit=4):
  url = f'https://modelmarket.dev/ai-market/v2/search?intent={intent}&limit={limit}'
  return json.load(urllib.request.urlopen(url, timeout=20)).get('matches') or []

found = {}
for q in ('murmuration', 'ablation', 'landauer', 'colony', 'physics'):
  hits = search(q)
  found[q] = [m.get('capability_id') for m in hits]
  print('—', q)
  for cid in found[q]:
    print(' ', cid)
assert any(found.values()), 'expected at least one physics-related capability'
```

```python
# Cell 2: optional live invoke of a free/public physics-ish cap if listed.
flat = [cid for cids in found.values() for cid in cids if cid]
print('candidates', flat[:5])
if flat:
  # Prefer search metadata over blind invoke — paid caps need a channel.
  pick = flat[0]
  meta = next(m for m in search(pick.split('.')[0], limit=8) if m.get('capability_id') == pick)
  print('picked', pick, 'price', meta.get('price_per_call_usd'))
  print('description:', (meta.get('description') or '')[:160])
print('Academy: swarm / cascade / Landauer labs with offline checks')
```
