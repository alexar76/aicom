```python
# AIMarket School — Dual-Orbit Seal
# Cell 1: pull live well-known hybrid envelope (do not invent PQ success).
import urllib.request, json

wk = json.load(urllib.request.urlopen('https://modelmarket.dev/.well-known/ai-market.json', timeout=20))
sig = wk.get('signature') or {}
print('classical', sig.get('algorithm'), 'pk', (sig.get('public_key') or '')[:20] + '…')
print('pq', sig.get('pq_algorithm'), 'has_pq_value', bool(sig.get('pq_value')))
print('pq_public_key bytes', len(sig.get('pq_public_key') or ''))
hybrid = bool(sig.get('algorithm') and sig.get('pq_algorithm') and sig.get('value') and sig.get('pq_value'))
print('hybrid_envelope', hybrid)
assert hybrid, 'expected live hybrid signature on hub well-known'
```

```python
# Cell 2: gate rule — refuse single-orbit receipts (teaching assertion).
def refuse_single_orbit(envelope):
  s = envelope or {}
  classical = bool(s.get('algorithm') and s.get('value'))
  pq = bool(s.get('pq_algorithm') and s.get('pq_value'))
  if classical and pq: return 'accept'
  return 'refuse'
print('gate live', refuse_single_orbit(sig))
print('gate classical-only', refuse_single_orbit({'algorithm': 'ed25519', 'value': 'x'}))
print('Academy: sign dual orbits locally + verify offline')
```
