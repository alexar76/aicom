```python
# Cell 2: minimum field policy before cryptographic verification.
def check_signature_fields(envelope):
  s = envelope or {}
  classical = bool(s.get('algorithm') and s.get('value'))
  pq = bool(s.get('pq_algorithm') and s.get('pq_value'))
  if classical and pq: return 'both present'
  return 'missing required fields'
print('live fields', check_signature_fields(sig))
print('classical-only fields', check_signature_fields({'algorithm': 'ed25519', 'value': 'x'}))
print('Academy: sign dual orbits locally + verify offline')
```
