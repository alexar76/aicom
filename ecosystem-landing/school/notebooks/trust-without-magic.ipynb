```python
# AIMarket School — Trust Without Magic
# Cell 1: hub identity + federated supply (measurable, not vibes).
import urllib.request, json

def get(url):
  with urllib.request.urlopen(url, timeout=20) as r:
    return json.load(r)

wk = get('https://modelmarket.dev/.well-known/ai-market.json')
print('hub', wk.get('name'), 'v' + str(wk.get('hub_version')))
print('local caps', wk.get('capabilities_count'))
print('federated caps', wk.get('federated_capabilities_count'))
print('payment_configured', wk.get('payment_configured'))
peers = wk.get('peers') or []
print('peers advertised', len(peers))
for p in peers[:5]:
  print(' ', p.get('name') or p.get('url'), 'trust', p.get('trust_score'), 'caps', p.get('capabilities_count'))
```

```python
# Cell 2: live stats if published — trust is a graph you can query.
try:
  st = get('https://modelmarket.dev/ai-market/v2/stats/live')
  print('stats/live keys', sorted(st)[:12] if isinstance(st, dict) else type(st))
  print(json.dumps(st, indent=2)[:600])
except Exception as e:
  print('stats/live:', e)
  print('still useful: well-known peers + federated_capabilities_count')
```
