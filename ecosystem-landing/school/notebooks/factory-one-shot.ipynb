```python
# AIMarket School — Idea → Product
# Cell 1: hub manifest — indexed tools the factory / agents can hire.
import urllib.request, json

man = json.load(urllib.request.urlopen('https://modelmarket.dev/ai-market/v2/manifest', timeout=20))
tools = man.get('tools') or man.get('capabilities') or []
print('tools indexed', man.get('total_capabilities') or len(tools))
print('local', man.get('local_capabilities'), 'federated', man.get('federated_capabilities'))
print('sample:')
for t in (tools[:5] if isinstance(tools, list) else []):
  print(' ', t.get('id') or t.get('capability_id') or t)
```

```python
# Cell 2: factory public surface (human storefront side of the pipeline).
for url in (
  'https://magic-ai-factory.com/api/health',
  'https://magic-ai-factory.com/health',
  'https://modelmarket.dev/ai-market/v2/stats/live',
):
  try:
    with urllib.request.urlopen(url, timeout=15) as r:
      body = r.read(400).decode('utf-8', 'replace')
    print(url, '→', r.status, body[:120].replace('\n',' '))
  except Exception as e:
    print(url, '→', type(e).__name__, e)
print('Academy: walk research→design→build→test→ship in the factory course')
```
