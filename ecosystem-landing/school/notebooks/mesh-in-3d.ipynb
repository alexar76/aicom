```python
# AIMarket School — See the Mesh in 3D
# Cell 1: live mesh endpoints.
import urllib.request, json

print('Live mesh: https://service-mesh.modelmarket.dev/')
print('Landing:   https://alexar76.github.io/ai-service-mesh/')

def try_json(url):
  try:
    with urllib.request.urlopen(url, timeout=15) as r:
      return r.status, json.load(r)
  except Exception as e:
    return None, str(e)

for path in ('/v1/stats', '/v1/topology', '/health', '/api/stats'):
  status, body = try_json('https://service-mesh.modelmarket.dev' + path)
  if status:
    keys = list(body)[:8] if isinstance(body, dict) else type(body)
    print(path, '→', status, keys)
  else:
    print(path, '→', body)
```

```python
# Cell 2: hub view of mesh-adjacent caps (topology / routing language).
data = json.load(urllib.request.urlopen(
  'https://modelmarket.dev/ai-market/v2/search?intent=mesh&limit=6', timeout=20))
for m in data.get('matches') or []:
  print(m.get('capability_id'), (m.get('description') or '')[:70])
print('Academy: probe alien-monitor health + map scenes in R3F')
```
