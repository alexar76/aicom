```python
# AIMarket School — Random You Can't Rig
# Cell 1: pull a signed chaos-VRF draw from Platon (public oracle path).
import urllib.request, json, urllib.error

INVOKE = 'https://oracles.modelmarket.dev/family/ai-market/v2/invoke'

def invoke(capability_id, payload=None):
    body = json.dumps({'capability_id': capability_id, 'input': payload or {}}).encode()
    req = urllib.request.Request(INVOKE, data=body, headers={'content-type': 'application/json'})
    with urllib.request.urlopen(req, timeout=30) as r:
        return json.load(r)

out = invoke('platon.random@v1')
output = out.get('output') or out
print('random_hex:', (output.get('random_hex') or '')[:64], '…')
proof = output.get('proof') or {}
print('scheme:', proof.get('scheme'))
print('keys in output:', sorted(output)[:12])
```

```python
# Cell 2: two draws must differ — bias would collapse entropy.
a = (invoke('platon.random@v1').get('output') or {}).get('random_hex')
b = (invoke('platon.random@v1').get('output') or {}).get('random_hex')
print('draw A', a[:32] if a else None)
print('draw B', b[:32] if b else None)
assert a and b and a != b, 'successive VRF draws should differ'
print('ok: unbiasable draws differ; Academy verifies proofs offline')
```
