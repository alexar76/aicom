```python
# AIMarket School — Hubs That Knock
# Cell 1: hub well-known + federated capability count.
import urllib.request, json

HUB = 'https://modelmarket.dev'
wk = json.load(urllib.request.urlopen(HUB + '/.well-known/ai-market.json', timeout=20))
print('hub', wk.get('name'), 'v', wk.get('hub_version'))
print('federated caps', wk.get('federated_capabilities_count'))
print('signer', (wk.get('signer_public_key') or '')[:24] + '…')
```

```python
# Cell 2: federation peers — compare declared story vs crawled counts.
peers = json.load(urllib.request.urlopen(HUB + '/ai-market/v2/federation/peers', timeout=25)).get('peers') or []
print('peers', len(peers))
for p in peers[:6]:
  decl = (p.get('declared') or {})
  print(f'  {p.get('url')}  status={p.get('status')}  caps={p.get('capabilities_count')}  trust={p.get('trust_score')}')
  print(f'    declared id={decl.get('id')} hub_version={decl.get('hub_version')}')
assert peers, 'expected at least one federated peer'
print('Academy: announce → preview quarantine → approve without vibe-trust')
```
