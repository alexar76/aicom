```python
# AIMarket School — Optimization With Receipts
# Cell 1: discover proof-bearing optimizer capabilities.
import urllib.request, json

url = 'https://modelmarket.dev/ai-market/v2/search?intent=optimize&limit=10'
data = json.load(urllib.request.urlopen(url, timeout=20))
matches = data.get('matches') or []
print(f'{len(matches)} optimize-related caps')
for m in matches:
  print('-', m.get('capability_id'))
  print(' ', (m.get('description') or '')[:100])
```

```python
# Cell 2: broaden the search — TSP / transport / proof language.
for intent in ('tsp', 'transport', 'proof', 'colony'):
  u = f'https://modelmarket.dev/ai-market/v2/search?intent={intent}&limit=4'
  hits = json.load(urllib.request.urlopen(u, timeout=20)).get('matches') or []
  print(f'— {intent}: {len(hits)}')
  for m in hits[:3]:
    print(' ', m.get('capability_id'))
print('Academy: call an optimizer and verify the receipt offline')
```
